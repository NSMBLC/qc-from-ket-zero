# Warm-up 0a: Sets, Functions, and Bit Strings
**Why this matters for the course:** *Basics of Quantum Information* lists sets and functions
among its recommended background, and its own first page already leans on that vocabulary
without defining it: a system is defined by its classical state *set* Σ, and operations on it
are *functions*. This warm-up follows the same sequence as Hunter's *An Introduction to Real
Analysis* (Ch. 1, IBM's own recommended reading): what a set and a function are, then two
special kinds of function, *onto*, *one-to-one*, and the *indicator function* χ_A: X → {0,1},
because measurement and reversible gates later in the course lean on exactly these ideas.
Cartesian products come next (also Hunter's own Ch. 1, §1.1.3), and Lab 2 builds composite
systems from them. Last is bit strings and indexing, the one thing here that isn't in
Hunter's chapter: the practical skill of knowing which entry of a numbered list belongs to
which state, built directly on the indicator-function idea from section 2. That skill
prevents the single most common stumble in all of quantum programming.

**⚠️ One notation clash to know about ahead of time:** if you read further into Hunter's
chapter, §1.5 defines Σ as the set of *all infinite binary sequences*, a different, unrelated
object from the Σ we use here (the *finite* classical state set of a bit, `{0,1}`). Same
Greek letter, two different meanings, in two different sections. Section 4 doesn't use Σ at
all for exactly this reason.

## ⚠️ One warning to file away for later: Qiskit's qubit ordering

When you reach multi-qubit work in Labs 2 and 3, you will meet a convention clash. The
course (and most textbooks) writes the state of qubits (A, B) with A's bit first: "10" means
A=1, B=0. Qiskit numbers qubits the other way around: qubit 0 is the RIGHTMOST bit of the
string (little-endian). The same physical state can therefore print as "10" in one convention
and "01" in the other.

Nothing to do about it today. Just plant the flag: when a two-qubit result someday looks
"backwards", come back to this cell before doubting your math. It will be the ordering, and
your `bits_to_index` and `index_to_bits` are exactly the tools to untangle it.
